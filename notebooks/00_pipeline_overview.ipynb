# 00 — AlphaFold 2 from Scratch: pipeline overview

Use this notebook as a map before opening the implementation notebooks. Each stage has one responsibility, one primary Python file, and one detailed notebook.

## End-to-end map

```text
A3M alignment
     |
     v
[1 Feature extraction] --> numerical feature batch
     |
     v
[2 Feature embedding]  --> m (MSA), z (pairs), e (extra MSA)
     |
     v
[3 Evoformer]          --> refined m and z
     |
     v
[5 Structure module]   --> residue positions and orientations
     |
     v
C-alpha structure + distance and confidence predictions

[4 Geometry] supplies transforms and local-coordinate operations to
             the structure module, teacher loader, and geometric losses.
```

The notebook numbers follow conceptual dependency order. Geometry appears before the structure module because understanding local frames makes invariant point attention much easier.

## Where each idea lives

| Stage | Notebook | Python file | Input | Output |
|---|---|---|---|---|
| Feature extraction | `01_feature_extraction.ipynb` | `feature_extraction.py` | A3M text | feature dictionary |
| Feature embedding | `02_feature_embedding.ipynb` | `feature_embedding.py` | feature dictionary | `m`, `z`, `e` |
| Evoformer | `03_evoformer.ipynb` | `evoformer.py` | `m`, `z`, `e` | refined `m`, `z` |
| Geometry | `04_geometry.ipynb` | `geometry.py` | points/transforms | geometric operations |
| Structure module | `05_structure_module.ipynb` | `structure_module.py` | `s`, `z` | frames and C-alpha positions |
| Full model | `06_full_model.ipynb` | `model.py` | feature batch | all predictions |

Training is deliberately separated from architecture: use notebooks 07 and 08 only after the six core stages are clear.

## Names to remember

```text
S = selected MSA rows       E = extra MSA rows       R = residues

m  (S,R,c_m)    sequence-alignment representation
z  (R,R,c_z)    residue-pair representation
e  (E,R,c_e)    cheaper extra-MSA representation
s  (R,c_s)      one vector per query residue
T  (R,4,4)      residue position and orientation
ca (R,3)        predicted C-alpha coordinates
```
