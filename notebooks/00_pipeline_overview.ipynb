# 00: AlphaFold 2 from Scratch pipeline overview

This notebook gives an overview of the model. Each stage links to one Python file and one detailed notebook.


## End-to-end map

```text
Related protein sequences (A3M alignment)
     |
     v
[1 Feature extraction] --> model-ready numbers
     |
     v
[2 Feature embedding]  --> sequence, residue-pair, and extra-MSA representations
     |
     v
[3 Evoformer]          --> refined sequence and residue-pair information
     |
     v
[5 Structure module]   --> residue positions and orientations
     |
     v
3D C-alpha backbone trace + distance and confidence predictions

[4 Geometry] provides the rotation, translation, and coordinate tools
             used to predict and evaluate the structure.
```

Geometry comes before the structure module because local coordinate frames are needed to understand invariant point attention.


## Where each idea lives

| Stage | Notebook | Python file | Input | Output |
|---|---|---|---|---|
| Feature extraction | `01_feature_extraction.ipynb` | `feature_extraction.py` | Related sequences in A3M format | Named input tensors |
| Feature embedding | `02_feature_embedding.ipynb` | `feature_embedding.py` | Input tensors | Initial sequence, residue-pair, and extra-MSA representations |
| Evoformer | `03_evoformer.ipynb` | `evoformer.py` | Sequence and pair representations | Refined sequence and pair information |
| Geometry | `04_geometry.ipynb` | `geometry.py` | Points and transformations | Rotation, translation, and coordinate operations |
| Structure module | `05_structure_module.ipynb` | `structure_module.py` | Per-residue and pair information | Residue frames and a 3D C-alpha backbone trace |
| Full model | `06_full_model.ipynb` | `model.py` | Extracted features | Backbone trace, distance predictions, and confidence |

Training is separate from the architecture. Continue to notebooks 07 and 08 after the six core stages are clear.


## Names to remember

```text
S = selected related sequences
E = additional related sequences
R = amino-acid residues

m  (S, R, c_m)    information about every residue in each selected sequence
z  (R, R, c_z)    information about every pair of residues
e  (E, R, c_e)    a smaller representation of additional sequences
s  (R, c_s)       one structure-ready vector per query residue
T  (R, 4, 4)      each residue's 3D position and orientation
ca (R, 3)         one C-alpha position per residue, forming a backbone trace
```

The C-alpha trace is a simplified representation of the protein backbone, not a complete all-atom structure.
